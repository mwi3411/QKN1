```python
#Inspection of data set
df.info()
```

```python
#change of column headers to standard format
df.rename(columns={"Business ID": "business_id","Business State":"business_state","Total Long-term Debt":"total_long_term_debt","Total Equity":"total_equity","Debt to Equity":"debt_to_equity","Total Liabilities":"total_liabilities","Total Revenue":"total_revenue"}, inplace=True)
```

```python
df.head()
```

```python
# determine negative values in debt_to_equity
neg_values = df['debt_to_equity'] < 0
print(df[neg_values])
```

```python
# The code below will identify duplicates
# Identify duplicates in 'column_name' and sum them
dubplicates_in_business_id = df.duplicated(subset=['business_id']).sum()
print(f"Number of duplicates in business_id from data set: {dubplicates_in_business_id}")
```
